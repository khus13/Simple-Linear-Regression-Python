```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv(r"C:\Users\Khus\OneDrive\Documents\Python Projects\Simple Linear Regression\canada_per_capita_income.csv")
df
```

```python
%matplotlib inline
```

```python
plt.scatter(df["year"], df["per capita income (US$)"])
plt.xlabel("year")
plt.ylabel("per capita income (US$)")
plt.show()
```

```python
reg = linear_model.LinearRegression() #function that creates a linear regression model
reg.fit(df[['year']], df['per capita income (US$)']) #Fitting the data means training the linear regression model using the available data points
```

```python
reg.predict(pd.DataFrame({'year': [2016]})) #Price prediction for the year 2016, Using pd.DataFrame since inputing year directly will raise errors because reg was initially seeded with 2D arrays, and the year field is 0D
```

```python
reg.coef_ #Shows the value of the Coefficient from formula y = mx + b, where y = vertical graph, m = the slope or gradient (coefficient), x = horizontal graph, b = y-intercept
```

```python
reg.intercept_
```
